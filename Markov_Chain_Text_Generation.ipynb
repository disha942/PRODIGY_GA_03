# Task-03: Text Generation with Markov Chains

## Objective
Implement a simple text generation algorithm using Markov Chains.

This project builds a statistical model that predicts the probability of the next word based on the previous word. The trained Markov Chain is then used to generate new text from a starting word.

### Technologies Used
- Python
- Markov Chains
- Random Probability Sampling
- Google Colab

### Workflow
1. Prepare sample text data
2. Build a Markov Chain model
3. Calculate word transition probabilities
4. Generate new text using the learned probabilities
5. Display the generated text

```python
import random
import re
from collections import defaultdict, Counter

print("Libraries imported successfully!")
```

```python
text = """
Artificial intelligence is transforming the world.
Machine learning is a powerful part of artificial intelligence.
Deep learning allows machines to learn complex patterns.
Generative AI can create text, images, music, and code.
Python is widely used for artificial intelligence and machine learning.
Learning artificial intelligence requires practice and experimentation.
Technology continues to change the way people learn and work.
"""

print(text)
```

```python
words = re.findall(r'\b\w+\b', text.lower())

print("Total words:", len(words))
print("First 20 words:", words[:20])
```

```python
markov_chain = defaultdict(Counter)

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    markov_chain[current_word][next_word] += 1

print("Markov Chain created successfully!")
```

```python
probabilities = {}

for current_word, next_words in markov_chain.items():
    total = sum(next_words.values())

    probabilities[current_word] = {
        word: count / total
        for word, count in next_words.items()
    }

print("Transition probabilities calculated successfully!")
```

```python
def generate_text(start_word, length=30):
    current_word = start_word.lower()
    generated_words = [current_word]

    for _ in range(length - 1):
        if current_word not in probabilities:
            break

        next_words = list(probabilities[current_word].keys())
        next_probabilities = list(probabilities[current_word].values())

        current_word = random.choices(
            next_words,
            weights=next_probabilities,
            k=1
        )[0]

        generated_words.append(current_word)

    return " ".join(generated_words)

print("Text generation function created successfully!")
```

```python
generated_text = generate_text("artificial", length=30)

print("Generated Text:")
print(generated_text)
```

```python
start_words = ["artificial", "machine", "technology", "python"]

for word in start_words:
    print(f"\nStarting word: {word}")
    print(generate_text(word, length=20))
```

```python
print("Sample Markov Chain Transitions:\n")

for word in list(markov_chain.keys())[:10]:
    print(f"{word} → {dict(markov_chain[word])}")
```

```python
start_word = input("Enter a starting word: ")

generated_text = generate_text(start_word, length=30)

print("\nGenerated Text:")
print(generated_text)
```

```python
print("===== MARKOV CHAIN MODEL SUMMARY =====")
print("Training words:", len(words))
print("Unique words:", len(set(words)))
print("States in Markov Chain:", len(markov_chain))
print("Model type: Word-level Markov Chain")
print("Generation method: Probability-based random sampling")
print("======================================")
```

## Conclusion

The model learns the statistical relationship between consecutive words in a training dataset and uses probability-based sampling to generate new text. Although a simple Markov Chain does not understand the meaning or context of language like modern language models, it provides a clear introduction to the fundamental concept of probabilistic text generation.
